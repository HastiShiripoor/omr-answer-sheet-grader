# OMR Answer Sheet Grader
Grades scanned answer sheets in `sample/sheets/` against the key `sample/answer_key.png`.
All logic lives in `omr_grader.py`.

```python
from pathlib import Path
import omr_grader as g

SHEETS = Path('sample/sheets')
KEY = Path('sample/answer_key.png')
OUT = Path('output')
```

## 1. Read the answer key
Green markers in the key image are the correct options (0 = no marker; question 149 is deleted).

```python
key = g.read_key(KEY)
print('first 20 key answers:', key[:20])
print('questions without a key marker:', [i + 1 for i, k in enumerate(key) if k == 0])
```

## 2. Detail table for one sheet

```python
sheet = sorted(SHEETS.glob('*.tif'))[0]
answers = g.read_sheet(sheet)
table = g.detail_table(answers, key)
print(sheet.name, '->', g.score_percent(g.grade(answers, key)), '%')
table.head(15)
```

## 3. Grade every sheet and save the CSV files

```python
status_df, score_df = g.grade_folder(SHEETS, KEY, OUT)
score_df
```
